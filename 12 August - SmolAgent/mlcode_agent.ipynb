```python
import os
import litellm
import gradio as gr
from smolagents import  CodeAgent , DuckDuckGoSearchTool ,LiteLLMModel
import google.generativeai as genai
```

```python
# Enable debugging (optional, useful for troubleshooting)
litellm._turn_on_debug()
```

```python
# Set your Google API key for the Generative AI API
os.environ["GOOGLE_API_KEY"] = "GOOGLE_API_KEY"  # Replace with your actual API key
genai.configure(api_key=os.environ["GOOGLE_API_KEY"])
```

```python
# Initialize Gemini Model for SmolAgent
model = LiteLLMModel(
    model_id ="gemono/gemini-2.5-flash",
    api_key = os.environ["GOOGLE_API_KEY"]
)
```

```python
#%pip install ddgs
```

```python
ml_code_agent = CodeAgent(
    tools=[DuckDuckGoSearchTool()],
    model=model,
    additional_authorized_imports=[
        "pandas",
        "numpy",
        "sklearn",
        "json"
    ]
)
```

```python
# Function to Generate ML code Step-by-Step

def generate_ml_code(prompt):
    full_prompt = """ Create a step-by-step plan to write python code for svm model using sklearn to classify the isis dataset"""

    try:
        response = ml_code_agent.run(full_prompt)
        return response
    except Exception as e:
        return f"An error occurred: {str(e)}"
```

```python
# Gradio UI for User Interaction
iface = gr.Interface(
    fn=generate_ml_code,
    inputs="text",
    outputs="text",
    title="ML Code Generator (Gemini 1.5 Flash)",
    description="Enter a Machine Learning coding task, and the AI will generate Python code using Gemini 1.5 Flash."
)
```

```python
# Launch Gradio App
iface.launch()
```
